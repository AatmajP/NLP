```python

from sklearn.feature_extraction.text import CountVectorizer

# Step 1: Define your text documents (the corpus)
corpus = [
    "The cat sat on the mat.",
    "The dog sat on the log.",
    "The cat chased the dog."
]

# Step 2: Initialize CountVectorizer
# This automatically handles lowercase conversion and strips punctuation

vectorizer = CountVectorizer(ngram_range=(1, 2))

# Step 3: Fit the vocabulary and transform text into numerical vectors
X = vectorizer.fit_transform(corpus)
print("vocabulary",vectorizer.get_feature_names_out())
print("X",X.toarray())


```
