```python

from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Define your corpus (collection of documents)
corpus = [
    "The cat sat on the mat.",
    "The dog played in the park.",
    "Cats and dogs are great pets."
]

# 2. Initialize the TF-IDF Vectorizer
# Note: scikit-learn uses a slightly modified formula by default to avoid division by zero.
vectorizer = TfidfVectorizer()

# 3. Fit and transform the corpus to calculate TF-IDF scores
tfidf_matrix = vectorizer.fit_transform(corpus)

# 4. Get the distinct feature names (words)
feature_names = vectorizer.get_feature_names_out()



print("vocab",feature_names)

print("bow",tfidf_matrix.toarray())
```
